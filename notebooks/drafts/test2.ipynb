```python
import pandas as pd
import requests
from io import StringIO
import numpy as np

url = 'https://docs.google.com/document/d/e/2PACX-1vSvM5gDlNvt7npYHhp_XfsJvuntUhq184By5xO_pA4b_gCWeXb6dM6ZxwN8rE6S4ghUsCj2VKR21oEP/pub'

# url = 'https://docs.google.com/document/d/e/2PACX-1vTMOmshQe8YvaRXi6gEPKKlsC6UpFJSMAk4mQjLm_u1gmHdVVTaeh7nBNFBRlui0sTZ-snGwZM4DBCT/pub'

def block_array(url):
  response = requests.get(url)
  parse_string = StringIO(response.content.decode('utf-8'))
  tables = pd.read_html(parse_string)

  arr = np.array(tables[0])
  x_coords = arr[1:, 0]
  y_coords = arr[1:, 2]

  x_coords.astype(int).max()
  y_coords.astype(int).max()
  values = arr[1:,1]

  arr_msg = np.zeros((x_coords.astype(int).max()+1, y_coords.astype(int).max()+1), dtype=object)
  arr_msg[x_coords.astype(int), y_coords.astype(int)] = values
  return arr_msg

arr2 = block_array(url)

for row in arr2:
    print("".join(str(value) if value != 0 else " " for value in row))
```
