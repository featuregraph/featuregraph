```python
%load_ext autoreload
%autoreload 2

import sys
import featuregraph as fg
import numpy as np

print("Python:", sys.executable)
print("FeatureGraph:", fg.__file__)
```

```python
bidmc = fg.datasets.bidmc(subject=1)
bidmc
```

```python
lag = 2
epsilon = 0.001
```

```python
respiration_numpy = bidmc["respiration"].to_numpy()
```

```python
respiration_numpy
```

```python
numpy_difference = np.diff(
    respiration_numpy,
    n=1,
    prepend=np.nan,
)
numpy_difference
```

```python
transition = fg.transition.Transition(
    signals="respiration",
    group="subject",
    diff_lag=1,
    eps=epsilon,
)
transition_features = transition.fit_transform(bidmc)

np.testing.assert_allclose(
    transition_features["respiration_difference"],
    numpy_difference,
    equal_nan=True,
)
```
